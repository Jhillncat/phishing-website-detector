```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

#Loading the UCI PhiUSIIL Phishing URL (Website) dataset
dataset = fetch_ucirepo(id=967)
X = dataset.data.features
y = dataset.data.targets

print("Features shape:",X.shape)
print("Target shape:",y.shape)
display(X.head())
display(y.head())
```

```python
print("Target column:")
print(y.columns)

print("\nClass counts:")
print(y.iloc[:, 0].value_counts())
```

```python
#Checking the dataset for missing values and duplicates

print("Missing values in X:")
print(X.isnull().sum().sum())

print("\nDuplicate rows in X:")
print(X.duplicated().sum())

print("\nData types:")
print(X.dtypes.value_counts())
```

```python
#Finding the columns that contain text

text_columns = X.select_dtypes(include="object").columns

print("Text columns:")
print(list(text_columns))
```

```python
#Removing text columns that we won't use for the first models

X_ml = X.drop(columns=["URL", "Domain", "TLD", "Title"])

print("Original features:",X.shape[1])
print("ML features:",X_ml.shape[1])
print("\nML-ready data shape:")
print(X_ml.shape)
```

```python
#Preparing the target labels

y_ml = y["label"]

print("Target values:")
print(y_ml.value_counts())

print("\nFeature shape:",X_ml.shape)
print("Target shape:",y_ml.shape)
```

```python
#Now splitting the data into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_ml,y_ml, test_size=0.20,random_state=42,stratify=y_ml)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("\nTraining labels:")
print(y_train.value_counts())
print("\nTesting labels:")
print(y_test.value_counts())
```

```python
#Now looking at all of the features we can use for the baseline

print("Available features:")
print(list(X_ml.columns))
```

```python
#Showing features by their column numbers

for i, column in enumerate(X_ml.columns):
    print(i, column)
```

```python
#Checking on some features I may use for the rule based baseline

baseline_features = [
    "IsDomainIP",
    "HasObfuscation",
    "HasPasswordField",
    "HasExternalFormSubmit",
    "NoOfURLRedirect",
    "NoOfExternalRef"
]

for feature in baseline_features:
    print("\n", feature)
    print(X_ml[feature].value_counts().head(10))
```

```python
#Creating a simple rule based phishing baseline

def rule_based_prediction(data):
    score = (
        (data["IsDomainIP"] == 1).astype(int)
        + (data["HasObfuscation"] == 1).astype(int)
        + (data["HasPasswordField"] == 1).astype(int)
        + (data["HasExternalFormSubmit"] == 1).astype(int)
        + (data["NoOfURLRedirect"] > 0).astype(int)
        + (data["NoOfExternalRef"] > 5).astype(int)
    )

    #1 = Acceptable legitimate, 0 = phishing

    predictions = (score < 2).astype(int)
    return predictions

#Test the baseline on the testing data
baseline_predictions = rule_based_prediction(X_test)

print("Baseline predictions:")
print(pd.Series(baseline_predictions).value_counts())



```

```python
#Evaluaing the rule based baseline

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

baseline_accuracy = accuracy_score(y_test, baseline_predictions)

#Use 0 as the positive class b/c 0 = phishing
baseline_precision = precision_score(y_test, baseline_predictions, pos_label=0)
baseline_recall = recall_score(y_test,baseline_predictions, pos_label=0)
baseline_f1 = f1_score(y_test, baseline_predictions,pos_label=0)

print("Rule-Based Baseline Results")
print("---------------------------")
print("Accuracy:",round(baseline_accuracy, 4))
print("Precision:", round(baseline_precision,4))
print("Recall:", round(baseline_recall,4))
print("F1-score:", round(baseline_f1, 4))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, baseline_predictions))
```

```python
#Training the first machine learning model

from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression(max_iter=1000,random_state=42)

logistic_model.fit(X_train,y_train)
logistic_predictions = logistic_model.predict(X_test)

print("Logistic Regression model trained successfully.")
print("Number of predictions:",len(logistic_predictions))
```

```python
#Extract URL based features for a new URL
from urllib.parse import urlparse
import re
import ipaddress

def extract_app_features(url):
    #adding a scheme if the user leaves it out
    if not url.startswith(("http://", "https://")):
        url = "http://" + url

    parsed = urlparse(url)
    domain = parsed.netloc.split("@")[-1].split(":")[0]
    path_and_query = parsed.path + parsed.query

    #URL len
    url_length = len(url)

    #Domain len
    domain_length = len(domain)

    #Checking whether domain is an IP address
    try:
        ipaddress.ip_address(domain)
        is_domain_ip = 1
    except ValueError:
        is_domain_ip = 0

    #top level domain len
    domain_parts = domain.split(".")
    tld = domain_parts[-1] if len(domain_parts) > 1 else ""
    tld_length = len(tld)

    #num of subdomains
    no_of_subdomain = max(len(domain_parts) - 2, 0)

    # Obfuscation/complications
    obfuscated_matches= re.findall(r"%[0-9A-Fa-f]{2}", url)
    no_of_obfuscated_char= len(obfuscated_matches)
    has_obfuscation =1 if no_of_obfuscated_char > 0 or "@"in url else 0

    obfuscation_ratio = (
        no_of_obfuscated_char / url_length
        if url_length > 0 else 0
    )

    #letters
    no_of_letters = sum(c.isalpha() for c in url)

    letter_ratio = (
        no_of_letters / url_length
        if url_length> 0 else 0
    )

    #special characters
    no_of_equals = url.count("=")
    no_of_qmark = url.count("?")
    no_of_ampersand = url.count("&")

    no_of_other_special = sum(
        not c.isalnum() and c not in "=?&"
        for c in url
    )

    #HTTPS
    is_https = 1 if parsed.scheme == "https" else 0

    #character continuation rate
    runs= re.findall(r"(.)\1+", url)
    continuation_rate = (
        max([len(run) for run in runs], default=1) / url_length
        if url_length > 0 else 0
    )

    return {
        "URLLength": url_length,
        "DomainLength": domain_length,
        "IsDomainIP": is_domain_ip,
        "CharContinuationRate": continuation_rate,
        "TLDLength": tld_length,
        "NoOfSubDomain": no_of_subdomain,
        "HasObfuscation": has_obfuscation,
        "NoOfObfuscatedChar": no_of_obfuscated_char,
        "ObfuscationRatio": obfuscation_ratio,
        "NoOfLettersInURL": no_of_letters,
        "LetterRatioInURL": letter_ratio,
        "NoOfEqualsInURL": no_of_equals,
        "NoOfQMarkInURL": no_of_qmark,
        "NoOfAmpersandInURL": no_of_ampersand,
        "NoOfOtherSpecialCharsInURL": no_of_other_special,
        "IsHTTPS": is_https
    }
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib
import re
import ipaddress
from urllib.parse import urlparse


#load model
model = joblib.load("phishing_app_model.pkl")
app_features = joblib.load("app_features.pkl")


#get features from the url
def extract_app_features(url):

    if not url.startswith(("http://", "https://")):
        url = "http://" + url

    parsed = urlparse(url)

    domain = parsed.netloc.split("@")[-1].split(":")[0]

    #domain length
    domain_length = len(domain)

    #check if domain is an ip
    try:
        ipaddress.ip_address(domain)
        is_domain_ip = 1
    except ValueError:
        is_domain_ip = 0

    #get tld info
    domain_parts = domain.split(".")

    tld = domain_parts[-1] if len(domain_parts) > 1 else ""

    tld_length = len(tld)

    #count subdomains
    no_of_subdomain = max(len(domain_parts) - 2, 0)

    #check for obfuscation/complication
    obfuscated_matches = re.findall(
        r"%[0-9A-Fa-f]{2}",
        url
    )

    no_of_obfuscated_char = len(obfuscated_matches)
    has_obfuscation = (
        1
        if no_of_obfuscated_char > 0 or "@" in url
        else 0
    )
    obfuscation_ratio = (
        no_of_obfuscated_char / len(url)
        if len(url) > 0
        else 0
    )

    #count url characters
    no_of_equals = url.count("=")
    no_of_qmark = url.count("?")
    no_of_ampersand = url.count("&")

    #check if https
    is_https = (
        1
        if parsed.scheme == "https"
        else 0
    )
    return {
        "DomainLength": domain_length,
        "IsDomainIP": is_domain_ip,
        "TLDLength": tld_length,
        "NoOfSubDomain": no_of_subdomain,
        "HasObfuscation": has_obfuscation,
        "NoOfObfuscatedChar": no_of_obfuscated_char,
        "ObfuscationRatio": obfuscation_ratio,
        "NoOfEqualsInURL": no_of_equals,
        "NoOfQMarkInURL": no_of_qmark,
        "NoOfAmpersandInURL": no_of_ampersand,
        "IsHTTPS": is_https
    }

#page setup
st.set_page_config(
    page_title="Phishing Website Detector",
    layout="centered"
)

#page title
st.title(" Phishing Website Detector")
st.write(
    "Enter a website URL below to analyze its "
    "phishing risk using a machine learning model."
)

st.info(
    "This application uses a Random Forest classifier "
    "trained on URL-based features from the PhiUSIIL "
    "phishing URL dataset."
)

#url input
url = st.text_input(
    "Enter Website URL",
    placeholder="https://example.com"
)

#analyze url
if st.button("Analyze Website", type="primary"):

    if not url.strip():
        st.warning("Please enter a website URL.")

    else:
        try:

            #get features / put into dataframe
            features = extract_app_features(url)
            input_data = pd.DataFrame(
                [features]
            )

            #keep features in the same order & make predics
            input_data = input_data[
                app_features
            ]
            prediction = model.predict(
                input_data
            )[0]

            #get prediction probabilities
            probabilities = model.predict_proba(
                input_data
            )[0]

            #get phishing probability
            phishing_index = list(
                model.classes_
            ).index(0)
            phishing_probability = (
                probabilities[phishing_index]
            )

            #show result
            st.subheader("Analysis Result")
            if prediction == 0:

                st.error(
                    "⚠️ PHISHING WEBSITE DETECTED"
                )

            else:

                st.success(
                    "WEBSITE CLASSIFIED AS LEGITIMATE"
                )

            #show probability
            st.metric(
                "Model-Estimated Phishing Probability",
                f"{phishing_probability * 100:.2f}%"
            )

            #set risk level
            if phishing_probability >= 0.70:
                st.error("Risk Level: HIGH")

            elif phishing_probability >= 0.40:

                st.warning("Risk Level: MEDIUM")

            else:
                st.success("Risk Level: LOW")

            #show features
            with st.expander(
                "View URL Features Analyzed"
            ):

                feature_display = pd.DataFrame({
                    "Feature": input_data.columns,
                    "Value": input_data.iloc[0].values
                })
                st.dataframe(
                    feature_display,
                    use_container_width=True,
                    hide_index=True
                )

            #disclaimers
            st.caption(
                "This application is a research prototype. "
                "Its prediction should not be treated as a "
                "guarantee that a website is safe or malicious."
            )

        except Exception as e:

            st.error(
                f"Unable to analyze this URL: {e}"
            )
```
